# Why Complex Multiplication Adds Angles

I want to explain something that took me a while to really see. When you multiply two complex numbers, the angles add. That's the claim. But *why* does that happen? Not the algebraic proof — I mean, why does it make geometric sense?

Let me walk through it the way it finally clicked for me.

---

## First: what does multiplication actually do to a number?

There are two kinds of multiplication that are worth separating in your head.

**Regular multiplication** — say, "times 2" — just stretches a number. It makes it bigger or smaller. If I have a vector pointing somewhere, multiplying by 2 makes it point in the same direction but twice as long. No rotation. Just scaling.

**Multiplying by $i$** — this is the weird one. Because $i^2 = -1$, multiplying by $i$ rotates a number 90 degrees counterclockwise. That's all it does. Do it four times and you're back where you started.

A complex number like $(3 + 4i)$ is *both* at once. It scales *and* it rotates. The real part scales, the imaginary part rotates. When you multiply by $(3 + 4i)$, you are saying: "make this 3 times bigger in the same direction, and also add the rotational effect of $4i$."

---

## The triangle picture

Every complex number is a triangle. The triangle has two legs — one along the real axis (the real part), one going straight up (the imaginary part) — and a hypotenuse that is the actual arrow, the thing with the angle and the length.

When I write $3 + 4i$, I am describing a triangle: 3 across, 4 up. The hypotenuse is $\sqrt{3^2 + 4^2} = 5$. The angle of that hypotenuse is $\arctan(4/3)$.

This framing matters. Multiplication is going to be about combining triangles.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyArrowPatch

# ── helper: draw a right-angle marker ──────────────────────────────────────
def right_angle_mark(ax, corner, d1, d2, size=0.12):
    u = d1 / np.linalg.norm(d1)
    v = d2 / np.linalg.norm(d2)
    pts = [corner + size*u, corner + size*u + size*v, corner + size*v, corner]
    xs = [p[0] for p in pts]
    ys = [p[1] for p in pts]
    ax.plot(xs, ys, color='#888', lw=1, zorder=4)

# ── helper: arc label ───────────────────────────────────────────────────────
def angle_arc(ax, center, angle, radius, color, label='', label_r_scale=1.3):
    thetas = np.linspace(0, angle, 80)
    ax.plot(center[0] + radius*np.cos(thetas),
            center[1] + radius*np.sin(thetas), color=color, lw=2)
    mid = angle / 2
    ax.text(center[0] + label_r_scale*radius*np.cos(mid),
            center[1] + label_r_scale*radius*np.sin(mid),
            label, fontsize=10, color=color, ha='center', va='center')

# ── palette ──────────────────────────────────────────────────────────────────
BLUE   = '#1565C0'
GREEN  = '#2E7D32'
RED    = '#C62828'
PURPLE = '#6A1B9A'
ORANGE = '#E65100'
```

## A simple case first: multiplying a real number by a complex one

Let's start gentle. Take the real number 4 (just a point on the real axis, no rotation at all). Multiply by $(3 + i)$:

$$4 \times (3 + i) = 12 + 4i$$

What happened? I made the number **3 times larger** (that is the real part, 3), and I **added a rotation component** (that is the $i$ part, which lifts the result off the real axis). The result $12 + 4i$ is a scaled-up version of the original triangle $(3 + i)$, with one leg at 12 and another at 4.

The key insight: the result is just a **rescaled version of the multiplier's triangle**. Same angle, bigger size.

Below I draw that out. The original number 4 is just a flat arrow. After multiplying by $(3 + i)$, we land at $12 + 4i$. The angle of $(3 + i)$ becomes the angle of the result.

```python
fig, ax = plt.subplots(figsize=(9, 5))
ax.set_xlim(-1, 14); ax.set_ylim(-1, 6)
ax.set_aspect('equal')
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.grid(True, alpha=0.25)
ax.set_title('4 × (3 + i) = 12 + 4i   — scaling keeps the angle', fontsize=14, fontweight='bold')
ax.set_xlabel('Real'); ax.set_ylabel('Imaginary')

# original: 4 on real axis
ax.annotate('', xy=(4,0), xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color='#888', lw=2.5))
ax.text(2, 0.2, '4  (original)', fontsize=11, color='#888')

# multiplier triangle: (3 + i) scaled up to show angle
mult = np.array([3, 1])
ax.annotate('', xy=mult, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color=BLUE, lw=2.5, linestyle='dashed'))
ax.text(mult[0]+0.1, mult[1]+0.15, '3 + i  (multiplier)', fontsize=11, color=BLUE)
ax.plot([3, 3], [0, 1], color=BLUE, lw=1, ls='--', alpha=0.5)
ax.plot([0, 3], [0, 0], color=BLUE, lw=1, ls='--', alpha=0.5)

# result: 12 + 4i
result = np.array([12, 4])
ax.annotate('', xy=result, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color=RED, lw=3))
ax.plot(*result, 'o', color=RED, ms=9, zorder=5)
ax.text(result[0]+0.2, result[1]+0.2, '12 + 4i  (result)', fontsize=11, color=RED, fontweight='bold')
ax.plot([12, 12], [0, 4], color=RED, lw=1, ls='--', alpha=0.4)
ax.plot([0, 12], [0, 0], color=RED, lw=1, ls='--', alpha=0.4)

# angle arcs — both multiplier and result have the same angle
th_mult   = np.arctan2(1, 3)
th_result = np.arctan2(4, 12)   # same angle, arctan(4/12) = arctan(1/3)
angle_arc(ax, np.array([0,0]), th_mult,   0.9,  BLUE, f'θ = {np.degrees(th_mult):.1f}°', 1.45)
angle_arc(ax, np.array([0,0]), th_result, 2.2,  RED,  '',   1.0)

ax.text(4, 3.8,
        'Same angle!\nJust 4× bigger.',
        fontsize=10, color=RED, ha='center',
        bbox=dict(fc='#FFEBEE', ec=RED, boxstyle='round,pad=0.4'))

plt.tight_layout()
plt.show()
```

## Now the real thing: multiplying two complex numbers

Let's do $(3 + 4i) \times (2 + 3i)$.

FOIL gives us four terms. I want to group them differently — not by algebraic term, but by what each group *does geometrically*:

$$
(3 + 4i)(2 + 3i) = \underbrace{(3 + 4i) \times 2}_{\text{scale the triangle}} + \underbrace{(3 + 4i) \times 3i}_{\text{rotate and scale the triangle}}
$$

- The **first piece**, $(3 + 4i) \times 2$, is just a bigger version of our original triangle. Same angle, twice the size.
- The **second piece**, $(3 + 4i) \times 3i$, is the original triangle **rotated 90° and scaled by 3**.

The final complex number is where those two pieces meet when you add them tip-to-tail.

This is the picture that is embedded from the original explanation:

![complex multiplication visualization](images/complex_mult.png)

The diagram below generates this from scratch so you can see exactly what is happening with numbers.

```python
# (3 + 4i) * (2 + 3i)
# = (3+4i)*2  +  (3+4i)*3i
# Part A: scale by 2  =>  6 + 8i
# Part B: multiply by 3i => 3i*(3+4i) = 9i + 12i^2 = -12 + 9i
# Sum: (6 + 8i) + (-12 + 9i) = -6 + 17i

z1 = np.array([3.0, 4.0])   # 3 + 4i
# Part A: z1 * 2
A  = z1 * 2                  # (6, 8)
# Part B: z1 * 3i  =>  rotate z1 by 90° then scale by 3
z1_rot90 = np.array([-z1[1], z1[0]])   # (-4, 3)
B  = z1_rot90 * 3            # (-12, 9)
result = A + B               # (-6, 17)

fig, ax = plt.subplots(figsize=(10, 10))
ax.set_xlim(-15, 12); ax.set_ylim(-2, 20)
ax.set_aspect('equal')
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.grid(True, alpha=0.2)
ax.set_title('$(3+4i)(2+3i)$: scale then rotate-and-scale, add tip-to-tail', fontsize=14, fontweight='bold')
ax.set_xlabel('Real'); ax.set_ylabel('Imaginary')

# Original triangle z1 (small, for reference)
ax.annotate('', xy=z1, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color='#888', lw=2, linestyle='dashed'))
ax.text(z1[0]+0.15, z1[1]+0.15, '$3+4i$  (original)', fontsize=10, color='#888')
ax.plot([z1[0], z1[0]], [0, z1[1]], color='#aaa', lw=1, ls=':')
ax.plot([0, z1[0]], [0, 0], color='#aaa', lw=1, ls=':')

# Part A: from origin to A  (scale ×2)
ax.annotate('', xy=A, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color=BLUE, lw=3))
ax.text(A[0]+0.3, A[1]-0.8, f'$(3+4i)\\times 2$\n$= {int(A[0])}+{int(A[1])}i$\n(same angle, ×2)',
        fontsize=10, color=BLUE,
        bbox=dict(fc='#E3F2FD', ec=BLUE, boxstyle='round,pad=0.3'))
ax.plot([A[0], A[0]], [0, A[1]], color=BLUE, lw=1, ls='--', alpha=0.4)

# Part B: from tip of A to result  (rotate 90° then ×3)
ax.annotate('', xy=result, xytext=A,
            arrowprops=dict(arrowstyle='->', color=GREEN, lw=3))
mid_B = A + B/2
ax.text(mid_B[0]-4.5, mid_B[1]-0.5,
        f'$(3+4i)\\times 3i$\n$= {int(B[0])}+{int(B[1])}i$\n(rotate 90°, ×3)',
        fontsize=10, color=GREEN,
        bbox=dict(fc='#E8F5E9', ec=GREEN, boxstyle='round,pad=0.3'))

# Result: from origin to result
ax.annotate('', xy=result, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color=RED, lw=3.5))
ax.plot(*result, 'o', color=RED, ms=10, zorder=6)
ax.text(result[0]-3.5, result[1]+0.3,
        f'Result: $({int(result[0])}+{int(result[1])}i)$',
        fontsize=11, color=RED, fontweight='bold',
        bbox=dict(fc='#FFEBEE', ec=RED, boxstyle='round,pad=0.4'))

# Right-angle marker where Part A ends and Part B begins
right_angle_mark(ax, A,
                 A/np.linalg.norm(A),
                 B/np.linalg.norm(B),
                 size=0.5)

# Dashed completion of parallelogram
ax.plot([B[0], result[0]], [B[1], result[1]], '--', color=BLUE, lw=1, alpha=0.4)
ax.plot([A[0], result[0]], [A[1], result[1]], '--', color=GREEN, lw=1, alpha=0.4)

# Angles
th_z1     = np.arctan2(z1[1], z1[0])
th_A      = np.arctan2(A[1], A[0])
th_result = np.arctan2(result[1], result[0])
angle_arc(ax, np.zeros(2), th_z1,     1.2, '#888',  f'θ₁={np.degrees(th_z1):.0f}°', 1.5)
angle_arc(ax, np.zeros(2), th_result, 2.4, RED,     f'θ_result={np.degrees(th_result):.0f}°', 1.4)

ax.text(-14, 18,
        'The final arrow (red)\nhas the angle of both\ntriangles combined.',
        fontsize=10, color=RED, style='italic',
        bbox=dict(fc='white', ec=RED, boxstyle='round,pad=0.4'))

plt.tight_layout()
plt.show()
print(f"z1        = 3+4i,  angle = {np.degrees(np.arctan2(4,3)):.1f}°,  length = {np.sqrt(3**2+4**2):.2f}")
print(f"z2        = 2+3i,  angle = {np.degrees(np.arctan2(3,2)):.1f}°,  length = {np.sqrt(2**2+3**2):.2f}")
print(f"result    = {int(result[0])}+{int(result[1])}i")
print(f"sum angle = {np.degrees(np.arctan2(4,3) + np.arctan2(3,2)):.1f}°")
print(f"result angle (direct) = {np.degrees(np.arctan2(result[1], result[0])):.1f}°")
```

## But *why* do the angles add? The similar triangles argument.

This is the part I kept glossing over and then suddenly it made sense.

The question is: after we add Part A and Part B tip-to-tail, how do we know the result arrow has angle $\theta_1 + \theta_2$?

Here is the argument. Look at two triangles:

1. The **original** triangle made by $z_1 = (3 + 4i)$ plopped on top of $z_2 = (2 + 3i)$. That gives a combined angle of $\theta_1 + \theta_2$.

2. The triangle made by **Part A** (scale $z_1$ by 2) plopped on top of **Part B** (rotate $z_1$ by 90° and scale by 3).

In case 1, you are combining the original two triangles to get the combined angle.  
In case 2, you are combining two *scaled copies* of those same triangles.

And here is the punchline from ninth-grade geometry: **similar triangles have the same angles**. A bigger copy of a triangle has all the same angles as the original. It is just larger.

So combining scaled-up copies of the same triangles produces the same angle as combining the originals. The angle does not care how big the triangles are — only what shape they are.

The picture below makes this very clear. I draw the small original version of plopping $z_1$ onto $z_2$, and then the large scaled version. The angles at every corner are identical.

```python
# ── Similar triangles: small version vs large scaled version ──────────────────
# We show:
#   Small: z2  then  z1 plopped on tip of z2  → angle θ1+θ2
#   Large: A (=z1*2)  then  B (=z1_rot90*3) plopped on tip of A  → same angle

z1 = np.array([3.0, 4.0])
z2 = np.array([2.0, 3.0])
th1 = np.arctan2(z1[1], z1[0])
th2 = np.arctan2(z2[1], z2[0])

# ── scale z2 to sit nicely at z1's tip for the "small" version
# "plop z1 onto z2": start at origin, go z2, then add a scaled z1
# To make it similar, we scale z1 by |z2| and rotate it by th2
z1_scaled_rotated = np.array([
    np.cos(th2)*z1[0] - np.sin(th2)*z1[1],
    np.sin(th2)*z1[0] + np.cos(th2)*z1[1]
])   # this IS the multiplication result

fig, axes = plt.subplots(1, 2, figsize=(16, 8))

for ax_idx, (ax, title, scale_factor) in enumerate(zip(
        axes,
        ['Original triangles plopped together\n(small — just to see the angles)',
         'Scaled copies plopped together\n(same angles, bigger)'],
        [1.0, 1.0])):

    ax.set_aspect('equal')
    ax.axhline(0, color='k', lw=0.7); ax.axvline(0, color='k', lw=0.7)
    ax.grid(True, alpha=0.2)
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_xlabel('Real'); ax.set_ylabel('Imaginary')

    if ax_idx == 0:
        # ── small picture: z2 from origin, then z1 from tip of z2
        arrow1 = z2
        arrow2_start = z2
        arrow2 = z1_scaled_rotated   # the rotated+scaled z1

        ax.set_xlim(-2, 12); ax.set_ylim(-1, 10)

        # z2 arrow
        ax.annotate('', xy=arrow1, xytext=(0,0),
                    arrowprops=dict(arrowstyle='->', color=GREEN, lw=2.8))
        ax.text(arrow1[0]+0.15, arrow1[1]-0.5, f'$z_2 = 2+3i$', fontsize=10, color=GREEN)

        # z1 plopped at tip of z2
        tip = arrow1 + arrow2
        ax.annotate('', xy=tip, xytext=arrow2_start,
                    arrowprops=dict(arrowstyle='->', color=BLUE, lw=2.8))
        ax.text(tip[0]+0.15, tip[1]+0.15, f'$z_1$ plopped on', fontsize=10, color=BLUE)

        # combined arrow
        ax.annotate('', xy=tip, xytext=(0,0),
                    arrowprops=dict(arrowstyle='->', color=RED, lw=3.5, linestyle='dashed'))
        ax.plot(*tip, 'o', color=RED, ms=9, zorder=6)
        ax.text(tip[0]-3, tip[1]-1.2, f'Combined angle = {np.degrees(th1+th2):.1f}°',
                fontsize=10, color=RED, fontweight='bold',
                bbox=dict(fc='#FFEBEE', ec=RED, boxstyle='round,pad=0.3'))

        angle_arc(ax, np.zeros(2), th2, 0.7, GREEN, f'θ₂={np.degrees(th2):.0f}°', 1.5)
        angle_arc(ax, np.zeros(2), th1+th2, 1.5, RED, f'θ₁+θ₂={np.degrees(th1+th2):.0f}°', 1.4)

        # mark angle at tip of z2
        angle_arc(ax, arrow2_start, th1, 0.5, BLUE, f'θ₁={np.degrees(th1):.0f}°', 1.8)

    else:
        # ── large picture: Part A, then Part B
        A_vec = z1 * 2
        B_vec = np.array([-z1[1], z1[0]]) * 3
        result_vec = A_vec + B_vec   # = (-6, 17)

        ax.set_xlim(-14, 10); ax.set_ylim(-1, 20)

        # Part A
        ax.annotate('', xy=A_vec, xytext=(0,0),
                    arrowprops=dict(arrowstyle='->', color=GREEN, lw=2.8))
        ax.text(A_vec[0]+0.2, A_vec[1]-0.8,
                f'Part A: $(3+4i)\\times 2$\n$= {int(A_vec[0])}+{int(A_vec[1])}i$',
                fontsize=10, color=GREEN,
                bbox=dict(fc='#E8F5E9', ec=GREEN, boxstyle='round,pad=0.3'))

        # Part B from tip of A
        ax.annotate('', xy=result_vec, xytext=A_vec,
                    arrowprops=dict(arrowstyle='->', color=BLUE, lw=2.8))
        mid = A_vec + B_vec/2
        ax.text(mid[0]-6, mid[1]-0.5,
                f'Part B: $(3+4i)\\times 3i$\n$= {int(B_vec[0])}+{int(B_vec[1])}i$',
                fontsize=10, color=BLUE,
                bbox=dict(fc='#E3F2FD', ec=BLUE, boxstyle='round,pad=0.3'))

        # Combined
        ax.annotate('', xy=result_vec, xytext=(0,0),
                    arrowprops=dict(arrowstyle='->', color=RED, lw=3.5, linestyle='dashed'))
        ax.plot(*result_vec, 'o', color=RED, ms=9, zorder=6)
        th_result = np.arctan2(result_vec[1], result_vec[0])
        ax.text(result_vec[0]-5, result_vec[1]+0.5,
                f'Same angle! = {np.degrees(th_result):.1f}°',
                fontsize=10, color=RED, fontweight='bold',
                bbox=dict(fc='#FFEBEE', ec=RED, boxstyle='round,pad=0.3'))

        angle_arc(ax, np.zeros(2), np.arctan2(A_vec[1], A_vec[0]), 1.5, GREEN,
                  f'θ₂={np.degrees(th2):.0f}°', 1.55)
        angle_arc(ax, np.zeros(2), th_result, 3.0, RED,
                  f'θ₁+θ₂={np.degrees(th_result):.0f}°', 1.3)
        angle_arc(ax, A_vec, np.arctan2(B_vec[1], B_vec[0]) - np.arctan2(A_vec[1],A_vec[0]),
                  1.2, BLUE, f'θ₁={np.degrees(th1):.0f}°', 2.2)

ax.text

plt.suptitle('Similar triangles: bigger copies, same angles', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
```

## How much did we rotate the original triangle, exactly?

When I multiply $(3 + 4i)$ by $(2 + 3i)$, I am rotating $(3 + 4i)$ by the angle of $(2 + 3i)$. That is the rotation.

The angle of $(2 + 3i)$ is:

$$\theta_2 = \arctan\!\left(\frac{3}{2}\right) \approx 56.3°$$

So the original triangle $(3 + 4i)$ — which sits at angle $\arctan(4/3) \approx 53.1°$ — gets rotated by $56.3°$, landing at roughly $109.4°$.

Let me check: $\arctan(17 / -6)$ in the second quadrant... yes, that is about $109.4°$. It works.

The rotation amount is always **the angle of the number you are multiplying by**. Not some mysterious formula — just the angle of the second number.

The picture below shows this explicitly: I draw $z_1$, then show it rotating by $\theta_2$ to land on the result.

```python
# Show z1 rotating by theta2 to reach the result
z1  = np.array([3.0, 4.0])
z2  = np.array([2.0, 3.0])
th1 = np.arctan2(z1[1], z1[0])
th2 = np.arctan2(z2[1], z2[0])

# Scale z1 by |z2| to get the magnitude of the result (for the arc)
r1 = np.linalg.norm(z1)
r2 = np.linalg.norm(z2)
r_result = r1 * r2   # = 5 * sqrt(13) ≈ 18.03

result = np.array([-6.0, 17.0])

fig, ax = plt.subplots(figsize=(9, 9))
ax.set_xlim(-8, 10); ax.set_ylim(-2, 20)
ax.set_aspect('equal')
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.grid(True, alpha=0.2)
ax.set_title(f'$z_1 = 3+4i$ rotates by θ₂ = {np.degrees(th2):.1f}° to reach the result',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Real'); ax.set_ylabel('Imaginary')

# z1
ax.annotate('', xy=z1, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color=BLUE, lw=3))
ax.text(z1[0]+0.2, z1[1]+0.15, f'$z_1 = 3+4i$\nangle = {np.degrees(th1):.1f}°',
        fontsize=10, color=BLUE, fontweight='bold')
angle_arc(ax, np.zeros(2), th1, 1.2, BLUE, f'{np.degrees(th1):.1f}°', 1.5)

# result
th_result = np.arctan2(result[1], result[0])
ax.annotate('', xy=result, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color=RED, lw=3))
ax.plot(*result, 'o', color=RED, ms=9, zorder=6)
ax.text(result[0]-4, result[1]+0.3,
        f'Result = $-6+17i$\nangle = {np.degrees(th_result):.1f}°',
        fontsize=10, color=RED, fontweight='bold')
angle_arc(ax, np.zeros(2), th_result, 3.5, RED, f'{np.degrees(th_result):.1f}°', 1.25)

# Curved arc showing the rotation from z1 to result, at the length of z1
arc_r = r1 + 0.4   # slightly outside z1's length
arc_angles = np.linspace(th1, th_result, 120)
ax.plot(arc_r * np.cos(arc_angles), arc_r * np.sin(arc_angles),
        color=ORANGE, lw=2.5, ls='--')
# Arrow at end of arc
da = arc_angles[-1] - arc_angles[-2]
ax.annotate('', xy=(arc_r*np.cos(arc_angles[-1]), arc_r*np.sin(arc_angles[-1])),
            xytext=(arc_r*np.cos(arc_angles[-2]), arc_r*np.sin(arc_angles[-2])),
            arrowprops=dict(arrowstyle='->', color=ORANGE, lw=2))

# Label the rotation arc
mid_arc = (th1 + th_result) / 2
ax.text(arc_r * 1.25 * np.cos(mid_arc), arc_r * 1.25 * np.sin(mid_arc),
        f'Rotate by\nθ₂ = {np.degrees(th2):.1f}°',
        fontsize=10, color=ORANGE, ha='center',
        bbox=dict(fc='#FFF3E0', ec=ORANGE, boxstyle='round,pad=0.3'))

# Also scale note
ax.text(4, 1,
        f'Length also scales:\n$|z_1| \\times |z_2| = {r1:.1f} \\times {r2:.2f} = {r_result:.2f}$',
        fontsize=10, color='#555',
        bbox=dict(fc='white', ec='#888', boxstyle='round,pad=0.4'))

plt.tight_layout()
plt.show()
```

## Why does the distance scale? The Pythagoras part.

When I multiply $(3 + 4i)$ by $(2 + 3i)$, the result is longer than either input. By how much?

Think about what happens geometrically. I scaled $(3 + 4i)$ by 2, getting a copy with length $2 \times |z_1|$. Then I scaled the 90°-rotated version by 3, getting another copy with length $3 \times |z_1|$. These two scaled copies sit at right angles to each other (because Part B is exactly perpendicular to Part A — one is the original direction, the other is rotated 90°).

Two arrows at right angles, with lengths $a \cdot |z_1|$ and $b \cdot |z_1|$ where $(a + bi)$ is the second number — what is the combined length?

Pythagoras: 

$$\text{combined} = \sqrt{(a \cdot |z_1|)^2 + (b \cdot |z_1|)^2} = |z_1| \cdot \sqrt{a^2 + b^2} = |z_1| \cdot |z_2|$$

So the length of the result is exactly $|z_1| \times |z_2|$. The lengths multiply.

This is why Pythagoras shows up — not because we are measuring a triangle *directly*, but because the two pieces (Part A and Part B) are **perpendicular** to each other. That right angle is what creates the Pythagorean situation.

Below I draw the right-angle geometry explicitly.

```python
# Pythagoras: Part A and Part B are perpendicular → combined length = |z1|*|z2|
z1 = np.array([3.0, 4.0])
A  = z1 * 2                                   # (6,  8),   length = |z1|*2
B  = np.array([-z1[1], z1[0]]) * 3           # (-12, 9),  length = |z1|*3
result = A + B                                # (-6, 17)

r1 = np.linalg.norm(z1)
r2 = np.linalg.norm(np.array([2.0, 3.0]))
r_A = np.linalg.norm(A)
r_B = np.linalg.norm(B)
r_result = np.linalg.norm(result)

fig, ax = plt.subplots(figsize=(10, 10))
ax.set_xlim(-15, 9); ax.set_ylim(-1, 20)
ax.set_aspect('equal')
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.grid(True, alpha=0.2)
ax.set_title('Part A ⊥ Part B → Pythagoras gives the combined length', fontsize=13, fontweight='bold')
ax.set_xlabel('Real'); ax.set_ylabel('Imaginary')

# Part A from origin
ax.annotate('', xy=A, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color=BLUE, lw=3))
ax.text(A[0]+0.3, A[1]/2,
        f'Part A\nlength = $|z_1|\\times 2 = {r_A:.1f}$',
        fontsize=10, color=BLUE,
        bbox=dict(fc='#E3F2FD', ec=BLUE, boxstyle='round,pad=0.3'))

# Part B from tip of A  (perpendicular)
ax.annotate('', xy=result, xytext=A,
            arrowprops=dict(arrowstyle='->', color=GREEN, lw=3))
mid_B = A + B/2
ax.text(mid_B[0]-6.5, mid_B[1],
        f'Part B\nlength = $|z_1|\\times 3 = {r_B:.1f}$\n(perpendicular to A)',
        fontsize=10, color=GREEN,
        bbox=dict(fc='#E8F5E9', ec=GREEN, boxstyle='round,pad=0.3'))

# Right-angle box at corner A
right_angle_mark(ax, A, A/np.linalg.norm(A), B/np.linalg.norm(B), size=0.6)

# Result (hypotenuse) — with brace annotation
ax.annotate('', xy=result, xytext=(0,0),
            arrowprops=dict(arrowstyle='->', color=RED, lw=3.5, linestyle='dashed'))
ax.plot(*result, 'o', color=RED, ms=9, zorder=6)
ax.text(result[0]-5, result[1]+0.5,
        f'Hypotenuse = result\nlength = $\\sqrt{{(|z_1|\\cdot 2)^2 + (|z_1|\\cdot 3)^2}}$\n'
        f'$= |z_1|\\cdot\\sqrt{{4+9}} = {r1:.0f}\\times{r2:.2f} = {r_result:.2f}$',
        fontsize=10, color=RED,
        bbox=dict(fc='#FFEBEE', ec=RED, boxstyle='round,pad=0.4'))

# Labels
ax.text(-14, 2,
        f'$|z_1| = {r1:.0f}$\n$|z_2| = \\sqrt{{4+9}} = {r2:.2f}$\n$|z_1 z_2| = {r_result:.2f}$',
        fontsize=11, family='monospace',
        bbox=dict(fc='white', ec='#888', boxstyle='round,pad=0.4'))

plt.tight_layout()
plt.show()
```

## Putting it all together: the one-paragraph summary

When I multiply $(3 + 4i)$ by $(2 + 3i)$, I am doing two things simultaneously:

1. I make a **scaled-up copy** of the original triangle (×2 for the real part) — same angle, bigger.
2. I make another **scaled and rotated copy** (×3 for the imaginary part, rotated 90°) — same angle relative to the first copy.

Plop those two copies tip-to-tail. Where they end up is the result.

The result has angle $\theta_1 + \theta_2$ because each copy keeps the same relative angles as the originals — that is just what "similar triangles" means. The result has length $|z_1| \times |z_2|$ because the two copies are perpendicular to each other, and Pythagoras applies.

No trigonometry. No hand-waving. Just triangles and algebra.

## Verify: a full interactive sweep

The diagram below sweeps $\theta_2$ from 0 to 360° and traces the result angle alongside $\theta_1 + \theta_2$, confirming they always agree.

```python
# Sweep theta2: confirm result angle = theta1 + theta2 always
z1 = 3 + 4j
th1 = np.angle(z1)

theta2_vals = np.linspace(0, 2*np.pi, 360)
result_angles = []
predicted    = []

for th2 in theta2_vals:
    z2 = np.cos(th2) + 1j*np.sin(th2)   # unit circle — pure rotation
    prod = z1 * z2
    result_angles.append(np.angle(prod))
    predicted.append(th1 + th2)

result_angles = np.unwrap(result_angles)
predicted     = np.array(predicted)

fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(np.degrees(theta2_vals), np.degrees(result_angles),
        color=RED, lw=2.5, label='Actual result angle')
ax.plot(np.degrees(theta2_vals), np.degrees(predicted),
        color=BLUE, lw=1.5, ls='--', label=r'Predicted: θ₁ + θ₂')
ax.set_xlabel('θ₂  (degrees)', fontsize=12)
ax.set_ylabel('Result angle (degrees)', fontsize=12)
ax.set_title('Multiplying by $e^{i\\theta_2}$ always adds exactly θ₂ to the angle', fontsize=13, fontweight='bold')
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
