# Polynomial Division — The Rectangle Method

## An Intuitive Approach Using Area Models

Forget long division with polynomials. Let's **see** what division actually means — it's just finding the missing side of a rectangle.

## 1️⃣ The Key Insight: Division = Finding a Missing Side

When you compute:

$$\frac{12}{3} = 4$$

You're really asking:

> **"I have a rectangle with area 12 and height 3. What's the width?"**

That's literally it. Division is geometry.

| Known | Unknown |
|-------|--------|
| Area = 12 | Width = ? |
| Height = 3 | |

$$\text{Area} = \text{Height} \times \text{Width}$$

$$12 = 3 \times \boxed{4}$$

````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# Simple numeric example: 12 ÷ 3 = 4
fig, ax = plt.subplots(figsize=(6, 4))

rect = patches.Rectangle((0, 0), 4, 3, linewidth=2.5,
                          edgecolor='#2c3e50', facecolor='#85c1e9')
ax.add_patch(rect)

# Area label
ax.text(2, 1.5, 'Area = 12', fontsize=18, ha='center', va='center', fontweight='bold')

# Height label
ax.annotate('', xy=(-0.4, 3), xytext=(-0.4, 0),
            arrowprops=dict(arrowstyle='<->', color='#e74c3c', lw=2))
ax.text(-0.7, 1.5, 'Height\n= 3', fontsize=13, ha='center', va='center',
        color='#e74c3c', fontweight='bold')

# Width label
ax.annotate('', xy=(4, -0.4), xytext=(0, -0.4),
            arrowprops=dict(arrowstyle='<->', color='#27ae60', lw=2))
ax.text(2, -0.7, 'Width = ?  →  4', fontsize=13, ha='center', va='center',
        color='#27ae60', fontweight='bold')

ax.set_xlim(-1.5, 5.5)
ax.set_ylim(-1.2, 4)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Division = Finding the Missing Side', fontsize=15, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()
````

## 2️⃣ Same Idea, But With Polynomials

Now let's do the exact same thing, but instead of numbers, we use polynomials.

We want to compute:

$$\frac{x^2 + 3x + 2}{x + 1}$$

Translation into rectangle language:

| Known | Unknown |
|-------|--------|
| Area = $x^2 + 3x + 2$ | Width = ? |
| Height = $x + 1$ | |

> **"What width do I need so that a rectangle with height $(x+1)$ has area $x^2 + 3x + 2$?"**

Let's build it piece by piece.

## 3️⃣ Build the Rectangle — Column by Column

### First column: match the biggest piece $x^2$

To get $x^2$, we need $x \times x = x^2$.

So start a column of width $x$.

But our height is $x + 1$, not just $x$. So this column gives us:

$$x \cdot (x + 1) = x^2 + x$$

```
        ┌──────────┐
    x   │   x²     │
        │          │
   +1   │   x      │
        └──────────┘
             x
```

**So far we have:** $x^2 + x$

**We wanted:** $x^2 + 3x + 2$

**Still missing:** $(x^2 + 3x + 2) - (x^2 + x) = 2x + 2$

### Second column: match the leftover $2x + 2$

We need $2x + 2$. Can we get that from our height $(x + 1)$?

$$2 \cdot (x + 1) = 2x + 2 \checkmark$$

So add a column of width $2$:

```
        ┌──────────┬──────┐
    x   │   x²     │  2x  │
        │          │      │
   +1   │   x      │  2   │
        └──────────┴──────┘
             x         2
```

Leftover: $0$. We're done!

````python
# Visualize the area model for (x² + 3x + 2) ÷ (x + 1)
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# --- Panel 1: First column only ---
ax = axes[0]
# x² block
r1 = patches.Rectangle((0, 1), 3, 3, linewidth=2, edgecolor='black', facecolor='#85c1e9')
ax.add_patch(r1)
ax.text(1.5, 2.5, '$x^2$', fontsize=18, ha='center', va='center', fontweight='bold')

# x block
r2 = patches.Rectangle((0, 0), 3, 1, linewidth=2, edgecolor='black', facecolor='#82e0aa')
ax.add_patch(r2)
ax.text(1.5, 0.5, '$x$', fontsize=16, ha='center', va='center', fontweight='bold')

# Labels
ax.text(-0.3, 2.5, '$x$', fontsize=14, ha='right', va='center', color='#e74c3c', fontweight='bold')
ax.text(-0.3, 0.5, '$1$', fontsize=14, ha='right', va='center', color='#e74c3c', fontweight='bold')
ax.text(1.5, -0.3, '$x$', fontsize=14, ha='center', va='top', color='#27ae60', fontweight='bold')

# Height brace
ax.annotate('', xy=(-0.7, 4), xytext=(-0.7, 0),
            arrowprops=dict(arrowstyle='<->', color='#e74c3c', lw=2))
ax.text(-1.2, 2, '$x+1$', fontsize=12, ha='center', va='center', rotation=90,
        color='#e74c3c', fontweight='bold')

ax.set_xlim(-1.8, 6)
ax.set_ylim(-0.8, 4.8)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Step 1: First column\nGives $x^2 + x$', fontsize=13, fontweight='bold')

# --- Panel 2: Both columns ---
ax = axes[1]
# x² block
r1 = patches.Rectangle((0, 1), 3, 3, linewidth=2, edgecolor='black', facecolor='#85c1e9')
ax.add_patch(r1)
ax.text(1.5, 2.5, '$x^2$', fontsize=18, ha='center', va='center', fontweight='bold')

# x block (bottom-left)
r2 = patches.Rectangle((0, 0), 3, 1, linewidth=2, edgecolor='black', facecolor='#82e0aa')
ax.add_patch(r2)
ax.text(1.5, 0.5, '$x$', fontsize=16, ha='center', va='center', fontweight='bold')

# 2x block (top-right)
r3 = patches.Rectangle((3, 1), 1.5, 3, linewidth=2, edgecolor='black', facecolor='#f9e79f')
ax.add_patch(r3)
ax.text(3.75, 2.5, '$2x$', fontsize=16, ha='center', va='center', fontweight='bold')

# 2 block (bottom-right)
r4 = patches.Rectangle((3, 0), 1.5, 1, linewidth=2, edgecolor='black', facecolor='#f1948a')
ax.add_patch(r4)
ax.text(3.75, 0.5, '$2$', fontsize=16, ha='center', va='center', fontweight='bold')

# Labels
ax.text(-0.3, 2.5, '$x$', fontsize=14, ha='right', va='center', color='#e74c3c', fontweight='bold')
ax.text(-0.3, 0.5, '$1$', fontsize=14, ha='right', va='center', color='#e74c3c', fontweight='bold')
ax.text(1.5, -0.3, '$x$', fontsize=14, ha='center', va='top', color='#27ae60', fontweight='bold')
ax.text(3.75, -0.3, '$2$', fontsize=14, ha='center', va='top', color='#27ae60', fontweight='bold')

# Height brace
ax.annotate('', xy=(-0.7, 4), xytext=(-0.7, 0),
            arrowprops=dict(arrowstyle='<->', color='#e74c3c', lw=2))
ax.text(-1.2, 2, '$x+1$', fontsize=12, ha='center', va='center', rotation=90,
        color='#e74c3c', fontweight='bold')

# Width brace
ax.annotate('', xy=(4.5, -0.6), xytext=(0, -0.6),
            arrowprops=dict(arrowstyle='<->', color='#27ae60', lw=2))
ax.text(2.25, -0.9, '$x + 2$', fontsize=12, ha='center', va='top',
        color='#27ae60', fontweight='bold')

ax.set_xlim(-1.8, 6)
ax.set_ylim(-1.3, 4.8)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Step 2: Add second column\nGives $2x + 2$', fontsize=13, fontweight='bold')

# --- Panel 3: Check totals ---
ax = axes[2]
# Same rectangle with totals
r1 = patches.Rectangle((0, 1), 3, 3, linewidth=2, edgecolor='black', facecolor='#85c1e9')
ax.add_patch(r1)
ax.text(1.5, 2.5, '$x^2$', fontsize=18, ha='center', va='center', fontweight='bold')

r2 = patches.Rectangle((0, 0), 3, 1, linewidth=2, edgecolor='black', facecolor='#82e0aa')
ax.add_patch(r2)
ax.text(1.5, 0.5, '$x$', fontsize=16, ha='center', va='center', fontweight='bold')

r3 = patches.Rectangle((3, 1), 1.5, 3, linewidth=2, edgecolor='black', facecolor='#f9e79f')
ax.add_patch(r3)
ax.text(3.75, 2.5, '$2x$', fontsize=16, ha='center', va='center', fontweight='bold')

r4 = patches.Rectangle((3, 0), 1.5, 1, linewidth=2, edgecolor='black', facecolor='#f1948a')
ax.add_patch(r4)
ax.text(3.75, 0.5, '$2$', fontsize=16, ha='center', va='center', fontweight='bold')

# Total check
ax.text(2.25, 4.5, 'Total = $x^2 + x + 2x + 2 = x^2 + 3x + 2$ ✓',
        fontsize=12, ha='center', va='center', fontweight='bold',
        bbox=dict(boxstyle='round,pad=0.4', facecolor='#d5f5e3', edgecolor='#27ae60', lw=2))

ax.set_xlim(-1, 6)
ax.set_ylim(-0.5, 5.2)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Check: All pieces sum\nto $x^2 + 3x + 2$ ✓', fontsize=13, fontweight='bold')

plt.suptitle('Building the Rectangle for $(x^2 + 3x + 2) \div (x + 1)$',
             fontsize=16, fontweight='bold', y=1.03)
plt.tight_layout()
plt.show()
````

## 4️⃣ Read Off the Answer

The total width is:

$$x + 2$$

So:

$$\frac{x^2 + 3x + 2}{x + 1} = x + 2$$

No long division. No synthetic division. Just **building a rectangle** and reading off the width.

````python
# Verify with actual numbers — plug in x = 5
x = 5
numerator = x**2 + 3*x + 2
denominator = x + 1
result = numerator / denominator

print(f"Let x = {x}:")
print(f"  Numerator:   x² + 3x + 2 = {numerator}")
print(f"  Denominator: x + 1        = {denominator}")
print(f"  Division:    {numerator} ÷ {denominator} = {result}")
print(f"  Answer:      x + 2        = {x + 2}")
print(f"  Match? {result == x + 2} ✓")
````

## 6️⃣ What About Remainders?

Sometimes the rectangle doesn't close perfectly. Let's try:

$$\frac{x^2 + 3x + 5}{x + 1}$$

### First column: width $x$

$$x \cdot (x + 1) = x^2 + x$$

**Missing:** $(x^2 + 3x + 5) - (x^2 + x) = 2x + 5$

### Second column: width $2$

$$2 \cdot (x + 1) = 2x + 2$$

**Missing:** $(2x + 5) - (2x + 2) = 3$

Hmm — we still have $3$ left over, and it's a constant. We can't make a new column from height $(x+1)$ that gives exactly $3$.

That leftover $3$ is the **remainder**.

$$\frac{x^2 + 3x + 5}{x + 1} = x + 2 + \frac{3}{x + 1}$$

> **A remainder means the rectangle can't close perfectly — there's a little piece left over that doesn't tile with the height.**

````python
# Visualize the remainder case
fig, ax = plt.subplots(figsize=(9, 6))

# x² block
r1 = patches.Rectangle((0, 1), 3, 3, linewidth=2, edgecolor='black', facecolor='#85c1e9')
ax.add_patch(r1)
ax.text(1.5, 2.5, '$x^2$', fontsize=18, ha='center', va='center', fontweight='bold')

# x block (bottom-left)
r2 = patches.Rectangle((0, 0), 3, 1, linewidth=2, edgecolor='black', facecolor='#82e0aa')
ax.add_patch(r2)
ax.text(1.5, 0.5, '$x$', fontsize=16, ha='center', va='center', fontweight='bold')

# 2x block (top-right)
r3 = patches.Rectangle((3, 1), 1.5, 3, linewidth=2, edgecolor='black', facecolor='#f9e79f')
ax.add_patch(r3)
ax.text(3.75, 2.5, '$2x$', fontsize=16, ha='center', va='center', fontweight='bold')

# 2 block (bottom-right)
r4 = patches.Rectangle((3, 0), 1.5, 1, linewidth=2, edgecolor='black', facecolor='#f1948a')
ax.add_patch(r4)
ax.text(3.75, 0.5, '$2$', fontsize=16, ha='center', va='center', fontweight='bold')

# Remainder block — doesn't fit the full height!
r5 = patches.Rectangle((4.5, 0), 0.8, 0.8, linewidth=2.5,
                        edgecolor='#8e44ad', facecolor='#d7bde2', linestyle='--')
ax.add_patch(r5)
ax.text(4.9, 0.4, '$3$', fontsize=15, ha='center', va='center',
        fontweight='bold', color='#8e44ad')
ax.text(4.9, 1.3, 'remainder!', fontsize=10, ha='center', va='center',
        color='#8e44ad', fontstyle='italic', fontweight='bold')

# Height labels
ax.text(-0.3, 2.5, '$x$', fontsize=14, ha='right', va='center', color='#e74c3c', fontweight='bold')
ax.text(-0.3, 0.5, '$1$', fontsize=14, ha='right', va='center', color='#e74c3c', fontweight='bold')

# Width labels
ax.text(1.5, -0.3, '$x$', fontsize=14, ha='center', va='top', color='#27ae60', fontweight='bold')
ax.text(3.75, -0.3, '$2$', fontsize=14, ha='center', va='top', color='#27ae60', fontweight='bold')

# Height brace
ax.annotate('', xy=(-0.7, 4), xytext=(-0.7, 0),
            arrowprops=dict(arrowstyle='<->', color='#e74c3c', lw=2))
ax.text(-1.2, 2, '$x+1$', fontsize=12, ha='center', va='center', rotation=90,
        color='#e74c3c', fontweight='bold')

# Width brace (just the clean part)
ax.annotate('', xy=(4.5, -0.6), xytext=(0, -0.6),
            arrowprops=dict(arrowstyle='<->', color='#27ae60', lw=2))
ax.text(2.25, -0.9, '$x + 2$', fontsize=12, ha='center', va='top',
        color='#27ae60', fontweight='bold')

ax.set_xlim(-1.8, 6.5)
ax.set_ylim(-1.3, 4.8)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('$(x^2 + 3x + 5) \div (x + 1) = x + 2$ remainder $3$',
             fontsize=15, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()
````

````python
# Verify the remainder case with numbers — plug in x = 10
x = 10
numerator = x**2 + 3*x + 5
denominator = x + 1
quotient = x + 2
remainder = 3

print(f"Let x = {x}:")
print(f"  Numerator:   x² + 3x + 5 = {numerator}")
print(f"  Denominator: x + 1        = {denominator}")
print(f"  Quotient:    x + 2        = {quotient}")
print(f"  Remainder:                 = {remainder}")
print(f"  Check: {denominator} × {quotient} + {remainder} = {denominator * quotient + remainder}")
print(f"  Matches numerator? {denominator * quotient + remainder == numerator} ✓")
````

## 8️⃣ The General Recipe

Here's the pattern — works every time:

1. **Set up the rectangle.** Height = divisor. Area = dividend.
2. **First column.** Pick a width that matches the leading term of the area. Multiply through the full height. Subtract from the area.
3. **Next column.** Repeat with what's left over.
4. **Keep going** until the leftover has lower degree than the height.

If there's anything left at the end → that's the **remainder**.

$$\frac{\text{Area}}{\text{Height}} = \text{Width} + \frac{\text{Remainder}}{\text{Height}}$$

````python
# Summary diagram: the general picture
fig, ax = plt.subplots(figsize=(9, 6))

# The big rectangle
outer = patches.FancyBboxPatch((0.5, 0.5), 7, 4, boxstyle='round,pad=0.1',
                                linewidth=3, edgecolor='#2c3e50', facecolor='#eaf2f8')
ax.add_patch(outer)

# Column dividers
ax.plot([3.5, 3.5], [0.5, 4.5], 'k--', linewidth=1.5, alpha=0.5)
ax.plot([5.5, 5.5], [0.5, 4.5], 'k--', linewidth=1.5, alpha=0.5)

# Column labels
ax.text(2, 2.5, 'Column 1', fontsize=13, ha='center', va='center',
        fontstyle='italic', alpha=0.6)
ax.text(4.5, 2.5, 'Column 2', fontsize=13, ha='center', va='center',
        fontstyle='italic', alpha=0.6)
ax.text(6.5, 2.5, '...', fontsize=18, ha='center', va='center', alpha=0.6)

# Area label
ax.text(4, 5.2, 'Area = Polynomial (dividend)', fontsize=14,
        ha='center', va='center', fontweight='bold',
        bbox=dict(boxstyle='round,pad=0.3', facecolor='#85c1e9', edgecolor='#2c3e50'))

# Height label
ax.annotate('', xy=(-0.1, 4.5), xytext=(-0.1, 0.5),
            arrowprops=dict(arrowstyle='<->', color='#e74c3c', lw=2.5))
ax.text(-0.6, 2.5, 'Height\n= Divisor', fontsize=12, ha='center', va='center',
        rotation=90, color='#e74c3c', fontweight='bold')

# Width label
ax.annotate('', xy=(7.5, 0), xytext=(0.5, 0),
            arrowprops=dict(arrowstyle='<->', color='#27ae60', lw=2.5))
ax.text(4, -0.4, 'Width = Quotient (answer)', fontsize=12,
        ha='center', va='top', color='#27ae60', fontweight='bold')

ax.set_xlim(-1.5, 8.5)
ax.set_ylim(-1, 6)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Polynomial Division = Finding the Missing Side',
             fontsize=17, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
````

## 🔥 The Intuition (This is the Moment)

You are building a rectangle:

- **Height** = divisor $(x + 1)$
- **Area** = polynomial $(x^2 + 3x + 2)$
- **Width** = answer $(x + 2)$

Each column you add is one **term** of the answer. You keep adding columns until the area is filled — or until you can't fill it perfectly (remainder).

> **Polynomial division = finding the missing side of a rectangle, one column at a time.**
