# The Simpler Quadratic Formula

## An Intuitive Approach Using Midpoints and Distance

Forget memorizing the quadratic formula. Let's **derive it intuitively** using a beautiful geometric insight about products.

## 1️⃣ The Key Insight: Products as Squares Minus Squares

Here's a powerful fact: **any product of two numbers can be represented using their mean and distance from the mean.**

Take two numbers $r$ and $s$. Let:
- $m$ = their **midpoint** (mean) = $\frac{r + s}{2}$
- $d$ = their **distance** from the midpoint

Then:
$$r = m - d \quad \text{and} \quad s = m + d$$

So the product becomes:
$$r \cdot s = (m - d)(m + d) = m^2 - d^2$$

**The product of two numbers equals the square of their mean minus the square of their distance!**

### Example: 5 × 7

- Mean: $m = \frac{5 + 7}{2} = 6$
- Distance: $d = 1$ (both numbers are 1 away from 6)

$$5 \times 7 = (6-1)(6+1) = 6^2 - 1^2 = 36 - 1 = 35 \checkmark$$

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# Visualize 5 × 7 = 6² - 1²
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

# Left: The rectangle 5 × 7
ax1 = axes[0]
rect = patches.Rectangle((0, 0), 7, 5, linewidth=2, edgecolor='#3498db', facecolor='#85c1e9')
ax1.add_patch(rect)
ax1.set_xlim(-0.5, 8)
ax1.set_ylim(-0.5, 6)
ax1.set_aspect('equal')
ax1.set_title('Rectangle: 5 × 7 = 35', fontsize=14, fontweight='bold')
ax1.text(3.5, 2.5, '35', fontsize=20, ha='center', va='center', fontweight='bold')
ax1.text(3.5, -0.3, '7 = 6 + 1', fontsize=11, ha='center')
ax1.text(-0.3, 2.5, '5 = 6 - 1', fontsize=11, ha='center', rotation=90)
ax1.axis('off')

# Middle: The square 6²
ax2 = axes[1]
square = patches.Rectangle((0, 0), 6, 6, linewidth=2, edgecolor='#27ae60', facecolor='#82e0aa')
ax2.add_patch(square)
ax2.set_xlim(-0.5, 7)
ax2.set_ylim(-0.5, 7)
ax2.set_aspect('equal')
ax2.set_title('Square: 6² = 36', fontsize=14, fontweight='bold')
ax2.text(3, 3, '36', fontsize=20, ha='center', va='center', fontweight='bold')
ax2.text(3, -0.3, 'm = 6', fontsize=11, ha='center')
ax2.axis('off')

# Right: The small square 1²
ax3 = axes[2]
small_square = patches.Rectangle((0, 0), 1, 1, linewidth=2, edgecolor='#e74c3c', facecolor='#f1948a')
ax3.add_patch(small_square)
ax3.set_xlim(-0.5, 7)
ax3.set_ylim(-0.5, 7)
ax3.set_aspect('equal')
ax3.set_title('Small Square: d² = 1² = 1', fontsize=14, fontweight='bold')
ax3.text(0.5, 0.5, '1', fontsize=16, ha='center', va='center', fontweight='bold')
ax3.text(0.5, -0.3, 'd = 1', fontsize=11, ha='center')
ax3.axis('off')

plt.suptitle('5 × 7 = 6² - 1² = 36 - 1 = 35', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 2️⃣ How Does This Help with Quadratics?

Consider the quadratic equation:
$$ax^2 + bx + c = 0$$

First, **normalize** by dividing by $a$:
$$x^2 + b'x + c' = 0$$

where $b' = \frac{b}{a}$ and $c' = \frac{c}{a}$.

If this quadratic has roots $r$ and $s$, we can write:
$$(x - r)(x - s) = 0$$

Expanding:
$$x^2 - (r + s)x + rs = 0$$

## 3️⃣ Three Key Facts

Comparing $x^2 + b'x + c' = 0$ with $x^2 - (r+s)x + rs = 0$:

| Coefficient | Equals | Meaning |
|-------------|--------|--------|
| $b'$ | $-(r + s)$ | Negative sum of roots |
| $c'$ | $rs$ | Product of roots |

Now here's the magic! Using our midpoint insight:

### Fact 1: The midpoint of the roots
$$m = \frac{r + s}{2} = \frac{-b'}{2}$$

### Fact 2: The product as difference of squares
$$c' = rs = (m - d)(m + d) = m^2 - d^2$$

### Fact 3: Solving for the distance
$$d^2 = m^2 - c'$$
$$d = \sqrt{m^2 - c'}$$

```python
# Visualize roots on a parabola with midpoint
fig, ax = plt.subplots(figsize=(10, 6))

# Example: roots at r=2 and s=5
r, s = 2, 5
m = (r + s) / 2  # midpoint = 3.5
d = (s - r) / 2  # distance = 1.5

x = np.linspace(0, 7, 100)
y = (x - r) * (x - s)

ax.plot(x, y, 'b-', linewidth=2, label=f'$(x-{r})(x-{s})$')
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)

# Mark roots
ax.plot([r, s], [0, 0], 'ro', markersize=12, label=f'Roots: r={r}, s={s}')

# Mark midpoint
ax.axvline(x=m, color='green', linestyle='--', linewidth=2, label=f'Midpoint m={m}')
ax.plot(m, 0, 'g^', markersize=15)

# Show distance d
ax.annotate('', xy=(s, -1), xytext=(m, -1),
            arrowprops=dict(arrowstyle='<->', color='purple', lw=2))
ax.text((m + s)/2, -1.5, f'd = {d}', fontsize=12, ha='center', color='purple', fontweight='bold')

ax.annotate('', xy=(r, -1), xytext=(m, -1),
            arrowprops=dict(arrowstyle='<->', color='purple', lw=2))
ax.text((m + r)/2, -1.5, f'd = {d}', fontsize=12, ha='center', color='purple', fontweight='bold')

ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_title(f'Roots are equidistant from the midpoint!\nr = m - d = {m} - {d} = {r}\ns = m + d = {m} + {d} = {s}', 
             fontsize=13, fontweight='bold')
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_ylim(-3, 5)

plt.tight_layout()
plt.show()
```

## 4️⃣ The Simpler Quadratic Formula

Given $x^2 + b'x + c' = 0$:

**Step 1:** Find the midpoint:
$$m = -\frac{b'}{2}$$

**Step 2:** Find the distance squared:
$$d^2 = m^2 - c'$$

**Step 3:** The roots are:
$$r, s = m \pm d = m \pm \sqrt{m^2 - c'}$$

That's it! **The roots are the midpoint plus or minus the distance.**

## 5️⃣ Example 1: $x^2 + 6x + 7 = 0$

Already normalized! So $b' = 6$ and $c' = 7$.

**Step 1:** Midpoint
$$m = -\frac{6}{2} = -3$$

**Step 2:** Distance squared
$$d^2 = (-3)^2 - 7 = 9 - 7 = 2$$
$$d = \sqrt{2}$$

**Step 3:** Roots
$$r, s = -3 \pm \sqrt{2}$$

So: $r = -3 - \sqrt{2} \approx -4.41$ and $s = -3 + \sqrt{2} \approx -1.59$

```python
import numpy as np
import matplotlib.pyplot as plt

# Example 1: x² + 6x + 7 = 0
b_prime, c_prime = 6, 7

# Step 1: Midpoint
m = -b_prime / 2
print(f"Step 1: Midpoint m = -{b_prime}/2 = {m}")

# Step 2: Distance squared
d_squared = m**2 - c_prime
d = np.sqrt(d_squared)
print(f"Step 2: d² = {m}² - {c_prime} = {m**2} - {c_prime} = {d_squared}")
print(f"        d = √{d_squared} = {d:.4f}")

# Step 3: Roots
r = m - d
s = m + d
print(f"Step 3: r = {m} - √{d_squared} = {r:.4f}")
print(f"        s = {m} + √{d_squared} = {s:.4f}")

# Verify
print(f"\nVerification: (x - r)(x - s) at x=0: {r * s:.4f} (should equal c' = {c_prime})")
print(f"              -(r + s) = {-(r + s):.4f} (should equal b' = {b_prime})")
```

```python
# Visualize Example 1
fig, ax = plt.subplots(figsize=(10, 6))

x = np.linspace(-6, 1, 200)
y = x**2 + 6*x + 7

ax.plot(x, y, 'b-', linewidth=2, label='$x^2 + 6x + 7$')
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)

# Mark roots
ax.plot([r, s], [0, 0], 'ro', markersize=12)
ax.annotate(f'r = -3 - √2\n≈ {r:.2f}', xy=(r, 0), xytext=(r-0.5, 2),
            fontsize=11, ha='center', arrowprops=dict(arrowstyle='->', color='red'))
ax.annotate(f's = -3 + √2\n≈ {s:.2f}', xy=(s, 0), xytext=(s+0.5, 2),
            fontsize=11, ha='center', arrowprops=dict(arrowstyle='->', color='red'))

# Mark midpoint
ax.axvline(x=m, color='green', linestyle='--', linewidth=2)
ax.plot(m, 0, 'g^', markersize=12)
ax.annotate(f'm = -3', xy=(m, 0), xytext=(m, -2), fontsize=12, ha='center', 
            color='green', fontweight='bold')

ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_title('Example 1: $x^2 + 6x + 7 = 0$\nRoots: $-3 \pm \sqrt{2}$', fontsize=14, fontweight='bold')
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_ylim(-3, 10)

plt.tight_layout()
plt.show()
```

## 6️⃣ Example 2: $3x^2 - 4x + 5 = 0$ (needs normalization)

First, divide by 3:
$$x^2 - \frac{4}{3}x + \frac{5}{3} = 0$$

So $b' = -\frac{4}{3}$ and $c' = \frac{5}{3}$.

**Step 1:** Midpoint
$$m = -\frac{-4/3}{2} = \frac{2}{3}$$

**Step 2:** Distance squared
$$d^2 = \left(\frac{2}{3}\right)^2 - \frac{5}{3} = \frac{4}{9} - \frac{15}{9} = -\frac{11}{9}$$

**d² is negative!** This means the roots are **complex numbers**.

**Step 3:** Roots
$$r, s = \frac{2}{3} \pm \sqrt{-\frac{11}{9}} = \frac{2}{3} \pm \frac{\sqrt{11}}{3}i$$

```python
# Example 2: 3x² - 4x + 5 = 0
a, b, c = 3, -4, 5

# Normalize
b_prime = b / a
c_prime = c / a
print(f"Normalized: x² + ({b_prime:.4f})x + {c_prime:.4f} = 0")

# Step 1: Midpoint
m = -b_prime / 2
print(f"\nStep 1: Midpoint m = {m:.4f}")

# Step 2: Distance squared
d_squared = m**2 - c_prime
print(f"Step 2: d² = {m:.4f}² - {c_prime:.4f} = {m**2:.4f} - {c_prime:.4f} = {d_squared:.4f}")

if d_squared < 0:
    print(f"\n⚠️  d² is NEGATIVE! The roots are complex.")
    d_imaginary = np.sqrt(-d_squared)
    print(f"\nStep 3: d = √({d_squared:.4f}) = {d_imaginary:.4f}i")
    print(f"        r = {m:.4f} - {d_imaginary:.4f}i")
    print(f"        s = {m:.4f} + {d_imaginary:.4f}i")
    
    # As fractions
    print(f"\nIn exact form: r, s = 2/3 ± (√11/3)i")
```

```python
# Visualize complex roots on the complex plane
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: The parabola (no real roots)
ax1 = axes[0]
x = np.linspace(-1, 3, 200)
y = 3*x**2 - 4*x + 5

ax1.plot(x, y, 'b-', linewidth=2, label='$3x^2 - 4x + 5$')
ax1.axhline(y=0, color='k', linewidth=0.5)
ax1.axvline(x=0, color='k', linewidth=0.5)
ax1.axvline(x=2/3, color='green', linestyle='--', linewidth=2, label=f'Vertex at x = 2/3')
ax1.set_xlabel('x', fontsize=12)
ax1.set_ylabel('y', fontsize=12)
ax1.set_title('Parabola never touches x-axis\n(No real roots)', fontsize=13, fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# Right: Complex plane
ax2 = axes[1]
m = 2/3
d_imag = np.sqrt(11)/3

ax2.axhline(y=0, color='k', linewidth=1)
ax2.axvline(x=0, color='k', linewidth=1)

# Plot complex roots
ax2.plot(m, d_imag, 'ro', markersize=12, label=f's = 2/3 + (√11/3)i')
ax2.plot(m, -d_imag, 'bo', markersize=12, label=f'r = 2/3 - (√11/3)i')

# Draw lines from origin
ax2.plot([0, m], [0, d_imag], 'r--', alpha=0.5)
ax2.plot([0, m], [0, -d_imag], 'b--', alpha=0.5)

# Show the distance from origin (modulus)
modulus = np.sqrt(m**2 + d_imag**2)
ax2.annotate(f'|z| = √(c\'/a) = √(5/3) ≈ {modulus:.2f}', 
             xy=(m/2, d_imag/2), fontsize=10, color='red')

ax2.set_xlabel('Real', fontsize=12)
ax2.set_ylabel('Imaginary', fontsize=12)
ax2.set_title('Complex Roots on the Complex Plane', fontsize=13, fontweight='bold')
ax2.legend(loc='upper right', fontsize=10)
ax2.grid(True, alpha=0.3)
ax2.set_xlim(-0.5, 2)
ax2.set_ylim(-1.5, 1.5)
ax2.set_aspect('equal')

plt.tight_layout()
plt.show()
```

## 7️⃣ Example 3: $x^2 - 6x + 10 = 0$ (Complex Roots)

Already normalized! So $b' = -6$ and $c' = 10$.

**Step 1:** Midpoint
$$m = -\frac{-6}{2} = 3$$

**Step 2:** Distance squared
$$d^2 = 3^2 - 10 = 9 - 10 = -1$$

**Step 3:** Roots
$$r, s = 3 \pm \sqrt{-1} = 3 \pm i$$

The roots are $3 + i$ and $3 - i$.

```python
# Example 3: x² - 6x + 10 = 0
b_prime, c_prime = -6, 10

m = -b_prime / 2
d_squared = m**2 - c_prime

print(f"Midpoint: m = {m}")
print(f"d² = {m}² - {c_prime} = {d_squared}")
print(f"\nRoots: {m} ± √{d_squared} = {m} ± i")
print(f"       r = {m} - i")
print(f"       s = {m} + i")

# The modulus
print(f"\nModulus |z| = √({m}² + 1²) = √{m**2 + 1} = √10 ≈ {np.sqrt(10):.3f}")
```

```python
# Visualize complex roots 3 ± i on the complex plane
fig, ax = plt.subplots(figsize=(8, 8))

ax.axhline(y=0, color='k', linewidth=1)
ax.axvline(x=0, color='k', linewidth=1)

# Plot the complex roots
ax.plot(3, 1, 'ro', markersize=15, label='s = 3 + i')
ax.plot(3, -1, 'bo', markersize=15, label='r = 3 - i')

# Draw the modulus (distance from origin)
theta = np.linspace(0, 2*np.pi, 100)
r_circle = np.sqrt(10)
ax.plot(r_circle * np.cos(theta), r_circle * np.sin(theta), 'g--', 
        alpha=0.5, linewidth=2, label=f'|z| = √10 ≈ {r_circle:.2f}')

# Lines from origin to roots
ax.plot([0, 3], [0, 1], 'r-', linewidth=2, alpha=0.7)
ax.plot([0, 3], [0, -1], 'b-', linewidth=2, alpha=0.7)

# Annotations
ax.annotate('√(3² + 1²) = √10', xy=(1.5, 0.7), fontsize=11, color='red', rotation=18)

ax.set_xlabel('Real', fontsize=14)
ax.set_ylabel('Imaginary', fontsize=14)
ax.set_title('Complex Roots of $x^2 - 6x + 10 = 0$\nRoots: $3 \pm i$', 
             fontsize=14, fontweight='bold')
ax.legend(loc='upper left', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_xlim(-1, 5)
ax.set_ylim(-3, 3)
ax.set_aspect('equal')

# Mark the real axis projection
ax.axvline(x=3, color='purple', linestyle=':', alpha=0.5)
ax.text(3.1, 2, 'Real part = m = 3', fontsize=10, color='purple')

plt.tight_layout()
plt.show()
```

## 8️⃣ Example 4: $x^2 - 7x + 12 = 0$ (Real Roots)

**Step 1:** Midpoint
$$m = -\frac{-7}{2} = \frac{7}{2} = 3.5$$

**Step 2:** Distance squared
$$d^2 = 3.5^2 - 12 = 12.25 - 12 = 0.25$$
$$d = 0.5$$

**Step 3:** Roots
$$r, s = 3.5 \pm 0.5$$
$$r = 3, \quad s = 4$$

**Check:** $(x-3)(x-4) = x^2 - 7x + 12$ ✓

```python
# Example 4: x² - 7x + 12 = 0
b_prime, c_prime = -7, 12

m = -b_prime / 2
d_squared = m**2 - c_prime
d = np.sqrt(d_squared)

print(f"Step 1: m = {m}")
print(f"Step 2: d² = {m}² - {c_prime} = {m**2} - {c_prime} = {d_squared}")
print(f"        d = {d}")
print(f"Step 3: r = {m} - {d} = {m - d}")
print(f"        s = {m} + {d} = {m + d}")

# Visualize
fig, ax = plt.subplots(figsize=(10, 6))

x = np.linspace(0, 7, 200)
y = x**2 - 7*x + 12

ax.plot(x, y, 'b-', linewidth=2, label='$x^2 - 7x + 12$')
ax.axhline(y=0, color='k', linewidth=0.5)

# Mark roots
ax.plot([3, 4], [0, 0], 'ro', markersize=12)
ax.annotate('r = 3', xy=(3, 0), xytext=(3, -1.5), fontsize=12, ha='center', fontweight='bold')
ax.annotate('s = 4', xy=(4, 0), xytext=(4, -1.5), fontsize=12, ha='center', fontweight='bold')

# Mark midpoint
ax.axvline(x=3.5, color='green', linestyle='--', linewidth=2)
ax.plot(3.5, 0, 'g^', markersize=12)
ax.annotate('m = 3.5', xy=(3.5, 0), xytext=(3.5, 1.5), fontsize=12, 
            ha='center', color='green', fontweight='bold')

ax.set_xlabel('x', fontsize=12)
ax.set_ylabel('y', fontsize=12)
ax.set_title('$x^2 - 7x + 12 = (x-3)(x-4)$\nRoots at 3 and 4, midpoint at 3.5', 
             fontsize=14, fontweight='bold')
ax.legend(loc='upper right', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_ylim(-3, 8)

plt.tight_layout()
plt.show()
```

## 9️⃣ The Three Cases

The value of $d^2 = m^2 - c'$ tells us everything:

| $d^2$ | $d$ | Roots | Parabola |
|-------|-----|-------|----------|
| $> 0$ | Real | Two distinct real roots | Crosses x-axis twice |
| $= 0$ | Zero | One repeated real root | Touches x-axis once |
| $< 0$ | Imaginary | Two complex conjugate roots | Never touches x-axis |

This $d^2$ is directly related to the **discriminant** from the traditional formula!

```python
# Visualize the three cases
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

x = np.linspace(-2, 6, 200)

# Case 1: d² > 0 (two real roots)
ax1 = axes[0]
y1 = x**2 - 4*x + 3  # roots at 1 and 3
ax1.plot(x, y1, 'b-', linewidth=2)
ax1.axhline(y=0, color='k', linewidth=0.5)
ax1.plot([1, 3], [0, 0], 'ro', markersize=10)
ax1.axvline(x=2, color='green', linestyle='--', linewidth=2)
ax1.set_title('$d^2 > 0$\nTwo Real Roots', fontsize=13, fontweight='bold', color='green')
ax1.set_xlabel('x')
ax1.grid(True, alpha=0.3)
ax1.text(2, 4, 'm=2, d²=1', fontsize=11, ha='center')

# Case 2: d² = 0 (one repeated root)
ax2 = axes[1]
y2 = x**2 - 4*x + 4  # root at 2 (repeated)
ax2.plot(x, y2, 'b-', linewidth=2)
ax2.axhline(y=0, color='k', linewidth=0.5)
ax2.plot([2], [0], 'ro', markersize=10)
ax2.axvline(x=2, color='green', linestyle='--', linewidth=2)
ax2.set_title('$d^2 = 0$\nOne Repeated Root', fontsize=13, fontweight='bold', color='orange')
ax2.set_xlabel('x')
ax2.grid(True, alpha=0.3)
ax2.text(2, 4, 'm=2, d²=0', fontsize=11, ha='center')

# Case 3: d² < 0 (complex roots)
ax3 = axes[2]
y3 = x**2 - 4*x + 5  # no real roots
ax3.plot(x, y3, 'b-', linewidth=2)
ax3.axhline(y=0, color='k', linewidth=0.5)
ax3.axvline(x=2, color='green', linestyle='--', linewidth=2)
ax3.set_title('$d^2 < 0$\nComplex Roots', fontsize=13, fontweight='bold', color='red')
ax3.set_xlabel('x')
ax3.grid(True, alpha=0.3)
ax3.text(2, 4, 'm=2, d²=-1', fontsize=11, ha='center')

for ax in axes:
    ax.set_ylim(-2, 8)
    ax.set_xlim(-1, 5)

plt.suptitle('The Three Cases Based on $d^2 = m^2 - c\'$', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 🔟 Connection to the Traditional Quadratic Formula

The traditional formula for $ax^2 + bx + c = 0$ is:
$$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

Let's see how our simpler formula relates:

For normalized $x^2 + b'x + c' = 0$ where $b' = b/a$ and $c' = c/a$:

$$m = -\frac{b'}{2} = -\frac{b}{2a}$$

$$d^2 = m^2 - c' = \frac{b^2}{4a^2} - \frac{c}{a} = \frac{b^2 - 4ac}{4a^2}$$

$$d = \frac{\sqrt{b^2 - 4ac}}{2a}$$

So:
$$x = m \pm d = -\frac{b}{2a} \pm \frac{\sqrt{b^2 - 4ac}}{2a} = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

**It's the same formula, just thought of differently!**

## 📝 Summary: The Simpler Quadratic Formula

For $x^2 + b'x + c' = 0$:

### Step 1: Find the midpoint
$$m = -\frac{b'}{2}$$

### Step 2: Find the distance squared
$$d^2 = m^2 - c'$$

### Step 3: The roots are
$$\boxed{r, s = m \pm \sqrt{d^2}}$$

---

**Why is this simpler?**
1. It's intuitive: roots are **midpoint ± distance**
2. You can visualize it on the parabola
3. The geometry makes complex roots natural: when $d^2 < 0$, you're measuring an "imaginary distance"
4. It separates the WHERE (midpoint) from the HOW FAR (distance)

```python
def solve_quadratic_simple(b_prime, c_prime, verbose=True):
    """
    Solve x² + b'x + c' = 0 using the simpler quadratic formula.
    
    Returns (r, s) - the two roots (may be complex).
    """
    # Step 1: Midpoint
    m = -b_prime / 2
    
    # Step 2: Distance squared
    d_squared = m**2 - c_prime
    
    # Step 3: Roots
    if d_squared >= 0:
        d = np.sqrt(d_squared)
        r = m - d
        s = m + d
    else:
        d_imag = np.sqrt(-d_squared)
        r = complex(m, -d_imag)
        s = complex(m, d_imag)
    
    if verbose:
        print(f"Equation: x² + ({b_prime})x + {c_prime} = 0")
        print(f"Midpoint: m = {m}")
        print(f"Distance²: d² = {d_squared}")
        print(f"Roots: r = {r}, s = {s}")
    
    return r, s

# Test it!
print("=" * 50)
solve_quadratic_simple(6, 7)      # x² + 6x + 7 = 0
print("=" * 50)
solve_quadratic_simple(-7, 12)    # x² - 7x + 12 = 0
print("=" * 50)
solve_quadratic_simple(-6, 10)    # x² - 6x + 10 = 0
```
