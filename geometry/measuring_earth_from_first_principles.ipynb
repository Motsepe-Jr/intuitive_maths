## 1. What the Sky Looks Like Before Theory

If we begin with only our eyes, the sky appears like a huge dome overhead.

```text
                     / - - - - - - \
                   /   visible sky   \
                  |       dome        |
==================[     observer     ]================== horizon
                  |   hidden half     |
                   \               /
                     \ - - - - - - /
```

Three observations matter:

- Stars do not wander randomly. Over hours, they trace smooth arcs.
- In the northern sky, many stars appear to turn around a nearly fixed point close to Polaris.
- Stars set below the western horizon and later rise again in the east.

That suggests the visible dome is only half of a larger idea: a full celestial sphere, with the horizon hiding the lower half from view.

````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

fig, ax = plt.subplots(figsize=(8, 5))
ax.set_aspect('equal')
ax.axis('off')

# Celestial sphere outline
sphere = patches.Circle((0, 0), 1, fill=False, linewidth=2)
ax.add_patch(sphere)

# Horizon and observer
ax.plot([-1.15, 1.15], [0, 0], linewidth=2)
ax.scatter([0], [0], s=70, zorder=3)
ax.text(0, -0.08, 'observer', ha='center', va='top')
ax.text(-1.18, 0.03, 'east', ha='right')
ax.text(1.18, 0.03, 'west', ha='left')

# Star paths as arcs around a tilted pivot
for radius in [0.35, 0.55, 0.75, 0.92]:
    arc = patches.Arc((-0.15, 0.35), radius * 2, radius * 0.9, angle=0, theta1=5, theta2=175, linewidth=1.6)
    ax.add_patch(arc)

ax.scatter([-0.15], [0.35], s=40)
ax.text(-0.15, 0.42, 'near Polaris', ha='center')
ax.text(0, 0.72, 'visible star arcs', ha='center')
ax.text(0, -0.68, 'hidden continuation below horizon', ha='center', color='gray')
ax.set_xlim(-1.35, 1.35)
ax.set_ylim(-1.1, 1.15)
plt.show()
````

## 2. The 12-Hour and 24-Hour Loop

If you see a group of stars at 11:00 PM, then roughly half a day later Earth has rotated about halfway around. If the Sun were not in the way, you would face a different part of the sky.

After one full stellar rotation, you face nearly the same stars again. More precisely, this takes one **sidereal day**, about 23 hours 56 minutes, not exactly one solar day of 24 hours.

The important intuition is:

$$
	ext{half turn} pprox 	ext{opposite sky}, quad 	ext{full turn} pprox 	ext{same sky again}.
$$

This supports the celestial-sphere model: stars seem to move on continuous circular tracks, while the horizon blocks half of the sphere from our view.

## 3. Why Model Earth as a Sphere?

Before measuring Earth, we need a reason to think Earth has a spherical shape.

A classic argument comes from lunar eclipses. During a lunar eclipse, Earth lies between the Sun and the Moon, so Earth's shadow falls on the Moon.

```text
[ Sun ]  ------------------>  [ Earth ]  ------------------>  shadow on Moon
```

The edge of Earth's shadow is always curved. A flat disk can cast a round-looking shadow only from special angles. From other angles it would look stretched or thin. A sphere, however, casts a circular shadow from every direction.

So the eclipse argument does not give the exact size of Earth, but it gives a powerful first-principles reason to use a sphere as the model.

## 4. Al-Biruni's First Move: Measure a Mountain

Al-Biruni's method begins with a practical problem: to measure Earth's radius from a horizon dip, we need to know the height of the observing point above sea level.

So first he measures a mountain height using similar triangles and a square board.

Let:

- $E$ be the mountain peak.
- $Z$ be the point directly below the peak at ground level.
- $G$ be the observation point on flat ground.
- $EZ$ be the unknown mountain height.
- $GE$ be the line-of-sight distance to the peak.

A simplified picture:

```text
          E peak
         /|
        / |
       /  |  EZ = mountain height
      /   |
     /    |
    G-----Z  ground
```

## 5. Similar Triangles on the Square Board

Al-Biruni uses a square board with side length $1$ meter. A small mark on the board gives a tiny triangle similar to the giant triangle pointing toward the mountain peak.

Given:

- Board side: $AD = DG = 1	ext{ m}$
- Small measured mark: $AT = 0.01648	ext{ cm} = 0.0001648	ext{ m}$

The similar-triangle ratio is:

$$
rac{DG}{GE} = rac{AT}{AD}.
$$

Since $DG = AD = 1$, this becomes:

$$
rac{1}{GE} = 0.0001648, quad GE = rac{1}{0.0001648} pprox 6068	ext{ m}.
$$

Now a second similar-triangle relation converts the line-of-sight distance $GE$ into the vertical mountain height $EZ$.

Given the measured ground drop:

$$
GH = 5.028	ext{ cm} = 0.05028	ext{ m},
$$

the corresponding ratio is:

$$
rac{EZ}{GE} = rac{GH}{DG}.
$$

````python
# Al-Biruni-style mountain height calculation
board_side_m = 1.0
AT_m = 0.0001648
GH_m = 0.05028

GE_m = board_side_m / AT_m
mountain_height_m = GE_m * (GH_m / board_side_m)

GE_m, mountain_height_m
````

The calculation gives:

$$
GE pprox 6068	ext{ m}, quad EZ pprox 305.1	ext{ m}.
$$

That means the mountain is about $305$ meters high, found without needing to measure straight down through the mountain.

## 6. Al-Biruni's Second Move: Measure the Horizon Dip

Now climb to the mountain peak. From height $h$, the visible horizon appears slightly below the local horizontal line.

Let:

- $O$ be Earth's center.
- $E$ be the observer at the mountain peak.
- $T$ be the horizon point where the line of sight just touches Earth.
- $r$ be Earth's radius.
- $h = 305.1	ext{ m}$ be the mountain height.
- $	heta = 34'$ be the dip angle of the horizon.

The line of sight $ET$ is tangent to Earth at $T$, so $OT erp ET$. This gives a right triangle:

```text
              E observer
             /|
            / | h
 horizon T /  |
          /   |
         /    | r + h
        O-----T
           r
```

The key trigonometric relation is:

$$
os(	heta) = rac{r}{r+h}.
$$

Solving for $r$:

$$
r = rac{hos(	heta)}{1 - os(	heta)}.
$$

````python
# Earth radius from horizon dip
h_m = mountain_height_m
theta_degrees = 34 / 60  # 34 arcminutes
theta_radians = np.deg2rad(theta_degrees)

earth_radius_m = h_m * np.cos(theta_radians) / (1 - np.cos(theta_radians))
earth_radius_km = earth_radius_m / 1000
earth_circumference_km = 2 * np.pi * earth_radius_km

modern_radius_km = 6371.0
modern_circumference_km = 40075.0
radius_error_percent = abs(earth_radius_km - modern_radius_km) / modern_radius_km * 100
circumference_error_percent = abs(earth_circumference_km - modern_circumference_km) / modern_circumference_km * 100

earth_radius_km, earth_circumference_km, radius_error_percent, circumference_error_percent
````

With the draft values:

$$
h pprox 305.1	ext{ m}, quad 	heta = 34' = 0.5667^irc.
$$

The estimate is roughly:

$$
r pprox 6238	ext{ km}.
$$

That is within about $2$ of the modern mean Earth radius of about $6371	ext{ km}$. The circumference estimate is:

$$
2i r pprox 39194	ext{ km},
$$

compared with Earth's equatorial circumference of about $40075	ext{ km}$.

## 7. What This Unlocks: Measuring Beyond Earth

Once Earth's radius is known, Earth itself becomes a measuring instrument.

For example, if two observers far apart on Earth measure the Moon's position against the background stars at the same time, the Moon appears shifted by a small angle. That shift is called **parallax**.

For small angles, the core idea is:

$$
	ext{distance} pprox rac{	ext{baseline}}{	ext{parallax angle in radians}}.
$$

The baseline can be a known distance across Earth, which depends on knowing Earth's size. So Al-Biruni's Earth measurement is not just about Earth. It helps turn the planet into a ruler for the sky.

## Summary

The complete chain is:

1. Star paths suggest a full celestial sphere, half hidden by the horizon.
2. Lunar eclipses strongly support modeling Earth as spherical.
3. Similar triangles let us measure a mountain height from a board and a few small distances.
4. The horizon dip from that mountain creates a right triangle with Earth's radius.
5. Solving $os(	heta) = r/(r+h)$ gives an estimate for Earth's radius.

The beauty of the method is that every step is visible: observations become geometry, geometry becomes a measurement, and the measurement becomes a model of the world.
