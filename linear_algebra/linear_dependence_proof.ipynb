# Linear Dependence — The Proof from First Principles

This note answers one question:

> **Why does the zero vector equation tell us whether vectors are dependent or independent?**

And along the way:
- What does "$= \vec{0}$" actually mean in the definition?
- Why do we care about the constants $c_1, c_2, \dots$ being zero or not?
- How does solving a system of equations prove dependence or independence?
- What is the geometric picture behind all of this?

We will build everything from scratch, no hand-waving.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

plt.rcParams['figure.figsize'] = (9, 6)
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.25
```

## 1) The core idea in plain language

You have a set of vectors. The question is:

> **Does every vector in this set add a new direction?**

If one vector can be built from the others (as a scaled copy, a sum, or any linear combination), then that vector is redundant. It does not add anything new. The set is **linearly dependent**.

If no vector can be built from the others, every vector genuinely contributes a new direction. The set is **linearly independent**.

That is the entire idea. Everything that follows is just a way to *test* for this.

## 2) Why we use the zero vector — building the test

Suppose you suspect that one vector, say $\mathbf{v}_1$, can be written as a combination of the others:

$$\mathbf{v}_1 = a_2\mathbf{v}_2 + a_3\mathbf{v}_3 + \cdots + a_n\mathbf{v}_n$$

Move everything to one side:

$$\mathbf{v}_1 - a_2\mathbf{v}_2 - a_3\mathbf{v}_3 - \cdots - a_n\mathbf{v}_n = \vec{0}$$

Now rename: let $c_1 = 1$, $c_2 = -a_2$, $c_3 = -a_3$, etc. Then:

$$c_1\mathbf{v}_1 + c_2\mathbf{v}_2 + c_3\mathbf{v}_3 + \cdots + c_n\mathbf{v}_n = \vec{0}$$

Notice: $c_1 = 1 \neq 0$, so **not all** the constants are zero.

This is exactly the formal definition of linear dependence.

---

The $\vec{0}$ is the **zero vector** — not the number zero. In $\mathbb{R}^2$ it is $\begin{bmatrix} 0 \\ 0 \end{bmatrix}$, in $\mathbb{R}^3$ it is $\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}$, and so on.

We set things equal to the zero vector because "one vector equals a combination of the others" is the same statement as "a combination of all of them equals zero" once you move terms across the equals sign.

## 3) The definition, stated cleanly

**Definition.** A set $\{\mathbf{v}_1, \mathbf{v}_2, \dots, \mathbf{v}_n\}$ is **linearly dependent** if there exist scalars $c_1, c_2, \dots, c_n$, **not all zero**, such that:

$$c_1\mathbf{v}_1 + c_2\mathbf{v}_2 + \cdots + c_n\mathbf{v}_n = \vec{0}$$

The set is **linearly independent** if the **only** solution is $c_1 = c_2 = \cdots = c_n = 0$ (the trivial solution).

---

Why does the trivial solution always exist? Because $0 \cdot \mathbf{v}_1 + 0 \cdot \mathbf{v}_2 + \cdots + 0 \cdot \mathbf{v}_n = \vec{0}$ is always true. Multiplying any vector by zero gives the zero vector. So that solution is boring — it tells you nothing.

The real question is: **is there another way to get the zero vector, where at least one constant is not zero?**

- If yes: the vectors are **dependent** (one cancels against the others).
- If no: the vectors are **independent** (the only way to get zero is to use no vector at all).

## 4) Why "not all zero" means one vector is built from the others

This is the proof that connects the definition back to the intuition.

Suppose we found constants $c_1, c_2, \dots, c_n$ (not all zero) such that:

$$c_1\mathbf{v}_1 + c_2\mathbf{v}_2 + \cdots + c_n\mathbf{v}_n = \vec{0}$$

At least one $c_i \neq 0$. Say $c_1 \neq 0$. Then isolate $\mathbf{v}_1$:

$$c_1\mathbf{v}_1 = -c_2\mathbf{v}_2 - c_3\mathbf{v}_3 - \cdots - c_n\mathbf{v}_n$$

Divide both sides by $c_1$ (legal because $c_1 \neq 0$):

$$\mathbf{v}_1 = -\frac{c_2}{c_1}\mathbf{v}_2 - \frac{c_3}{c_1}\mathbf{v}_3 - \cdots - \frac{c_n}{c_1}\mathbf{v}_n$$

There it is: $\mathbf{v}_1$ is a linear combination of the other vectors. It is redundant.

---

And the logic works in reverse too. If any vector can be written as a combination of the others, you can move everything to one side and get a non-trivial combination that equals $\vec{0}$.

So:

$$\text{non-trivial solution exists} \iff \text{one vector is a combination of the others}$$

The zero vector equation is not some arbitrary test. It is literally the statement "one vector equals the others" rearranged.

## 5) How do you actually check? Systems of equations

The equation $c_1\mathbf{v}_1 + c_2\mathbf{v}_2 + \cdots + c_n\mathbf{v}_n = \vec{0}$ is a vector equation. Each component gives you one scalar equation. So in $\mathbb{R}^2$ you get 2 equations, in $\mathbb{R}^3$ you get 3, and so on.

You solve this system. If the only solution is all constants zero, independent. If you find constants that are not all zero, dependent.

Let us work through both cases with concrete numbers.

## 6) Example 1 — Dependent vectors

Consider the set:

$$\left\{ \begin{bmatrix} 2 \\ 1 \end{bmatrix}, \begin{bmatrix} 3 \\ 2 \end{bmatrix}, \begin{bmatrix} 1 \\ 2 \end{bmatrix} \right\}$$

Three vectors in $\mathbb{R}^2$. Already suspicious — $\mathbb{R}^2$ is a plane, so at most 2 directions can be independent. A third vector in $\mathbb{R}^2$ must be buildable from the other two.

Set up the equation:

$$c_1 \begin{bmatrix} 2 \\ 1 \end{bmatrix} + c_2 \begin{bmatrix} 3 \\ 2 \end{bmatrix} + c_3 \begin{bmatrix} 1 \\ 2 \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}$$

Read off each component:

$$\text{Row 1: } 2c_1 + 3c_2 + c_3 = 0$$
$$\text{Row 2: } c_1 + 2c_2 + 2c_3 = 0$$

Two equations, three unknowns. More unknowns than equations — so there will be free variables (non-trivial solutions must exist).

### Solving the system by hand

From the two equations:

$$2c_1 + 3c_2 + c_3 = 0 \quad \text{...(i)}$$
$$c_1 + 2c_2 + 2c_3 = 0 \quad \text{...(ii)}$$

Multiply (ii) by 2:

$$2c_1 + 4c_2 + 4c_3 = 0 \quad \text{...(ii')}$$

Subtract (i) from (ii'):

$$(2c_1 + 4c_2 + 4c_3) - (2c_1 + 3c_2 + c_3) = 0$$
$$c_2 + 3c_3 = 0$$

So $c_2 = -3c_3$.

Let $c_3 = -1$ (free choice — any non-zero value works). Then $c_2 = 3$.

Substitute into (ii):

$$c_1 + 2(3) + 2(-1) = 0 \implies c_1 + 6 - 2 = 0 \implies c_1 = -4$$

**Solution:** $c_1 = -4, \quad c_2 = 3, \quad c_3 = -1$

None of these are zero. The vectors are **linearly dependent**.

```python
# Verify: -4*v1 + 3*v2 + (-1)*v3 should equal the zero vector
v1 = np.array([2, 1])
v2 = np.array([3, 2])
v3 = np.array([1, 2])

result = -4*v1 + 3*v2 + (-1)*v3
print(f'-4 * {v1} + 3 * {v2} + (-1) * {v3} = {result}')
print(f'Is it the zero vector? {np.allclose(result, 0)}')
```

### What does this tell us concretely?

From $-4\mathbf{v}_1 + 3\mathbf{v}_2 - \mathbf{v}_3 = \vec{0}$, isolate any one vector. For example, isolate $\mathbf{v}_1$:

$$-4\mathbf{v}_1 = -3\mathbf{v}_2 + \mathbf{v}_3$$
$$\mathbf{v}_1 = \frac{3}{4}\mathbf{v}_2 - \frac{1}{4}\mathbf{v}_3$$

So $\begin{bmatrix} 2 \\ 1 \end{bmatrix} = \frac{3}{4}\begin{bmatrix} 3 \\ 2 \end{bmatrix} - \frac{1}{4}\begin{bmatrix} 1 \\ 2 \end{bmatrix}$.

The first vector does not add a new direction — it is already "contained" in the other two.

```python
# Verify: v1 = (3/4)*v2 - (1/4)*v3
print(f'v1 = {v1}')
print(f'(3/4)*v2 - (1/4)*v3 = {(3/4)*v2 - (1/4)*v3}')
print(f'Match? {np.allclose(v1, (3/4)*v2 - (1/4)*v3)}')
```

```python
# Geometric picture: three vectors in R2, one is a combination of the other two

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# --- Left panel: all three vectors from the origin ---
ax = axes[0]
origin = np.array([0, 0])
vectors = [v1, v2, v3]
colors = ['#E53935', '#1E88E5', '#43A047']
labels = [r'$\mathbf{v}_1 = [2,1]$', r'$\mathbf{v}_2 = [3,2]$', r'$\mathbf{v}_3 = [1,2]$']

for v, c, lab in zip(vectors, colors, labels):
    ax.annotate('', xy=v, xytext=origin,
                arrowprops=dict(arrowstyle='->', color=c, lw=2.5))
    ax.text(v[0] + 0.1, v[1] + 0.15, lab, fontsize=13, color=c, fontweight='bold')

ax.set_xlim(-0.5, 4.5)
ax.set_ylim(-0.5, 3)
ax.set_aspect('equal')
ax.set_title('Three vectors in $\mathbb{R}^2$ — must be dependent', fontsize=14, fontweight='bold')
ax.axhline(0, color='black', lw=0.5)
ax.axvline(0, color='black', lw=0.5)

# --- Right panel: show v1 as a combination of v2 and v3 ---
ax = axes[1]

# Draw (3/4)*v2
scaled_v2 = (3/4) * v2
# Draw -(1/4)*v3 starting from the tip of (3/4)*v2
scaled_v3 = -(1/4) * v3

# (3/4)*v2 arrow
ax.annotate('', xy=scaled_v2, xytext=origin,
            arrowprops=dict(arrowstyle='->', color='#1E88E5', lw=2.5))
ax.text(scaled_v2[0]/2 - 0.3, scaled_v2[1]/2 + 0.2,
        r'$\frac{3}{4}\mathbf{v}_2$', fontsize=14, color='#1E88E5', fontweight='bold')

# -(1/4)*v3 arrow from tip of scaled_v2
ax.annotate('', xy=scaled_v2 + scaled_v3, xytext=scaled_v2,
            arrowprops=dict(arrowstyle='->', color='#43A047', lw=2.5))
ax.text(scaled_v2[0] + 0.15, scaled_v2[1] - 0.15,
        r'$-\frac{1}{4}\mathbf{v}_3$', fontsize=14, color='#43A047', fontweight='bold')

# v1 arrow (the result)
ax.annotate('', xy=v1, xytext=origin,
            arrowprops=dict(arrowstyle='->', color='#E53935', lw=2.5, linestyle='--'))
ax.text(v1[0] + 0.15, v1[1] - 0.25,
        r'$\mathbf{v}_1$', fontsize=14, color='#E53935', fontweight='bold')

ax.set_xlim(-0.5, 4.5)
ax.set_ylim(-1, 3)
ax.set_aspect('equal')
ax.set_title(r'$\mathbf{v}_1 = \frac{3}{4}\mathbf{v}_2 - \frac{1}{4}\mathbf{v}_3$ — redundant', fontsize=14, fontweight='bold')
ax.axhline(0, color='black', lw=0.5)
ax.axvline(0, color='black', lw=0.5)

plt.tight_layout()
plt.show()
```

## 7) Example 2 — Independent vectors

Now consider just two of those vectors:

$$\left\{ \begin{bmatrix} 2 \\ 1 \end{bmatrix}, \begin{bmatrix} 3 \\ 2 \end{bmatrix} \right\}$$

Set up the equation:

$$c_1 \begin{bmatrix} 2 \\ 1 \end{bmatrix} + c_2 \begin{bmatrix} 3 \\ 2 \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}$$

Component equations:

$$2c_1 + 3c_2 = 0 \quad \text{...(i)}$$
$$c_1 + 2c_2 = 0 \quad \text{...(ii)}$$

From (ii): $c_1 = -2c_2$. Substitute into (i):

$$2(-2c_2) + 3c_2 = 0 \implies -4c_2 + 3c_2 = 0 \implies -c_2 = 0 \implies c_2 = 0$$

And then $c_1 = -2(0) = 0$.

**The only solution is $c_1 = 0$ and $c_2 = 0$.** The trivial solution. No non-trivial combination of these vectors gives the zero vector.

Therefore: **linearly independent**.

### What does this mean?

The only way to combine $\begin{bmatrix} 2 \\ 1 \end{bmatrix}$ and $\begin{bmatrix} 3 \\ 2 \end{bmatrix}$ to get $\vec{0}$ is to multiply both by zero. That is, you cannot scale one to cancel the other. They point in genuinely different directions.

Neither vector is a multiple or combination of the other. Both are needed.

```python
# Verify with numpy: solve the system, only trivial solution
A = np.array([[2, 3],
              [1, 2]])

b = np.array([0, 0])
solution = np.linalg.solve(A, b)
print(f'Solution: c1 = {solution[0]}, c2 = {solution[1]}')
print(f'Only trivial solution exists => linearly independent')
```

```python
# Geometric picture: two independent vectors span the whole plane

fig, ax = plt.subplots(1, 1, figsize=(8, 7))

# Draw the two vectors
for v, c, lab in [(v1, '#E53935', r'$\mathbf{v}_1 = [2,1]$'),
                   (v2, '#1E88E5', r'$\mathbf{v}_2 = [3,2]$')]:
    ax.annotate('', xy=v, xytext=origin,
                arrowprops=dict(arrowstyle='->', color=c, lw=2.5))
    ax.text(v[0] + 0.1, v[1] + 0.15, lab, fontsize=13, color=c, fontweight='bold')

# Shade the parallelogram they span (shows they cover area, not just a line)
parallelogram = plt.Polygon([origin, v1, v1 + v2, v2], alpha=0.15, color='purple')
ax.add_patch(parallelogram)
ax.text((v1[0] + v2[0])/2, (v1[1] + v2[1])/2 + 0.15,
        'they span area\n(not collinear)', fontsize=11, ha='center',
        color='purple', fontstyle='italic')

# Extend lines through each vector to show they are not parallel
for v, c in [(v1, '#E53935'), (v2, '#1E88E5')]:
    t = np.linspace(-0.5, 2.5, 100)
    ax.plot(t * v[0], t * v[1], color=c, alpha=0.2, lw=1.5, linestyle='--')

ax.set_xlim(-1, 6)
ax.set_ylim(-0.5, 4.5)
ax.set_aspect('equal')
ax.axhline(0, color='black', lw=0.5)
ax.axvline(0, color='black', lw=0.5)
ax.set_title('Two independent vectors — different directions, span $\mathbb{R}^2$',
             fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 8) The determinant shortcut (for square cases)

When you have exactly $n$ vectors in $\mathbb{R}^n$, you can check dependence using the determinant of the matrix formed by the vectors.

For our two vectors in $\mathbb{R}^2$, form the matrix:

$$A = \begin{bmatrix} 2 & 3 \\ 1 & 2 \end{bmatrix}$$

$$\det(A) = 2 \cdot 2 - 3 \cdot 1 = 4 - 3 = 1 \neq 0$$

Determinant is not zero, so the only solution to $A\mathbf{c} = \vec{0}$ is $\mathbf{c} = \vec{0}$. Independent.

If $\det(A) = 0$, there would be non-trivial solutions, meaning dependent.

The determinant measures the "area" (or volume) spanned by the vectors. Zero area means they collapse onto a line (or hyperplane) — they are not pulling in enough distinct directions.

```python
# Determinant check
A = np.array([[2, 3],
              [1, 2]])
det = np.linalg.det(A)
print(f'det(A) = {det:.1f}')
print(f'Not zero => independent')
print()

# Compare: if we had two dependent vectors, e.g. v1 = [2,1] and w = [4,2] = 2*v1
B = np.array([[2, 4],
              [1, 2]])
det_B = np.linalg.det(B)
print(f'For dependent vectors [2,1] and [4,2]:')
print(f'det(B) = {det_B:.1f}')
print(f'Zero => dependent (one is a scalar multiple of the other)')
```

## 9) Putting it all together — the logical chain

Here is the full logical structure, from ground up:

**Step 1.** We want to know: can one vector be represented by the others?

**Step 2.** Rearranging that question gives: can a non-trivial linear combination equal $\vec{0}$?

**Step 3.** We write out $c_1\mathbf{v}_1 + c_2\mathbf{v}_2 + \cdots + c_n\mathbf{v}_n = \vec{0}$ and read off a system of equations (one equation per component).

**Step 4.** We solve the system.

- If the only solution is all $c_i = 0$: **independent**. The zero vector can only be reached by using none of the vectors. No vector is redundant.

- If a solution exists with some $c_i \neq 0$: **dependent**. We can isolate that vector and write it as a combination of the others. It is redundant.

**Step 5.** The reason "all $c_i = 0$" means independence is direct: multiplying a vector by zero gives the zero vector, so $0 \cdot \mathbf{v} = \vec{0}$ is not using $\mathbf{v}$ at all. It means we could not find any constants that cancel the vectors against each other. There is no redundancy.

## 10) A general-purpose function

Let us wrap the whole test into code, so you can experiment with any set of vectors.

```python
def check_linear_dependence(vectors):
    """
    Given a list of column vectors (each a 1-D numpy array of the same length),
    determine if they are linearly dependent or independent.

    Strategy: form the matrix whose columns are the vectors,
    then check the rank. If rank < number of vectors, dependent.
    """
    # Stack vectors as columns
    A = np.column_stack(vectors)
    rank = np.linalg.matrix_rank(A)
    n_vectors = len(vectors)

    print(f'Number of vectors : {n_vectors}')
    print(f'Dimension (R^d)   : {vectors[0].shape[0]}')
    print(f'Rank of matrix    : {rank}')
    print()

    if rank < n_vectors:
        print(f'Rank ({rank}) < number of vectors ({n_vectors})')
        print('=> LINEARLY DEPENDENT')
        print('   At least one vector can be written as a combination of the others.')
    else:
        print(f'Rank ({rank}) = number of vectors ({n_vectors})')
        print('=> LINEARLY INDEPENDENT')
        print('   No vector is redundant. Each adds a new direction.')


# Example 1: three vectors in R2 (dependent)
print('=== Example 1: {[2,1], [3,2], [1,2]} ===')
check_linear_dependence([np.array([2,1]), np.array([3,2]), np.array([1,2])])

print()
print('=== Example 2: {[2,1], [3,2]} ===')
check_linear_dependence([np.array([2,1]), np.array([3,2])])

print()
print('=== Example 3: {[1,0,0], [0,1,0], [0,0,1]} (standard basis in R3) ===')
check_linear_dependence([np.array([1,0,0]), np.array([0,1,0]), np.array([0,0,1])])

print()
print('=== Example 4: {[1,2], [2,4]} (one is a scalar multiple) ===')
check_linear_dependence([np.array([1,2]), np.array([2,4])])
```

## 11) The cancellation picture

Here is another way to see what is happening geometrically when vectors are dependent.

If $c_1\mathbf{v}_1 + c_2\mathbf{v}_2 + c_3\mathbf{v}_3 = \vec{0}$ with non-zero constants, it means the scaled vectors form a **closed loop** — you walk along one, then another, then another, and you end up exactly where you started (the origin).

They cancel. They close. That is dependence.

```python
# The cancellation: -4*v1 + 3*v2 + (-1)*v3 = 0
# Visualize the three scaled vectors forming a closed triangle (loop back to origin)

scaled = [(-4)*v1, 3*v2, (-1)*v3]
labels_sc = [r'$-4\mathbf{v}_1$', r'$3\mathbf{v}_2$', r'$-\mathbf{v}_3$']
colors_sc = ['#E53935', '#1E88E5', '#43A047']

fig, ax = plt.subplots(1, 1, figsize=(9, 7))

# Draw the chain of arrows: start at origin, each arrow starts where the previous ended
pos = np.array([0.0, 0.0])
positions = [pos.copy()]

for sv, c, lab in zip(scaled, colors_sc, labels_sc):
    next_pos = pos + sv
    ax.annotate('', xy=next_pos, xytext=pos,
                arrowprops=dict(arrowstyle='->', color=c, lw=3))
    mid = (pos + next_pos) / 2
    ax.text(mid[0] + 0.2, mid[1] + 0.3, lab, fontsize=14, color=c, fontweight='bold')
    pos = next_pos
    positions.append(pos.copy())

# Mark origin
ax.plot(0, 0, 'ko', markersize=8, zorder=5)
ax.text(0.2, -0.4, 'origin', fontsize=12, fontweight='bold')

# Mark that we returned to origin
ax.text(pos[0] - 0.2, pos[1] + 0.5,
        f'end = ({pos[0]:.0f}, {pos[1]:.0f})\nback to origin!',
        fontsize=11, color='black', fontstyle='italic',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

ax.set_aspect('equal')
ax.axhline(0, color='black', lw=0.5)
ax.axvline(0, color='black', lw=0.5)
ax.set_title('Dependent vectors: scaled versions form a closed loop back to the origin',
             fontsize=13, fontweight='bold')

# Auto-scale
all_pts = np.array(positions)
margin = 2
ax.set_xlim(all_pts[:,0].min() - margin, all_pts[:,0].max() + margin)
ax.set_ylim(all_pts[:,1].min() - margin, all_pts[:,1].max() + margin)

plt.tight_layout()
plt.show()
```

## Final takeaway

The zero vector equation $c_1\mathbf{v}_1 + c_2\mathbf{v}_2 + \cdots + c_n\mathbf{v}_n = \vec{0}$ is not an arbitrary trick. It is the direct algebraic translation of the question: "can one vector be built from the others?"

- If you *can* find non-zero constants that make the combination equal $\vec{0}$, the vectors cancel each other. One is redundant. **Dependent.**

- If the *only* way to reach $\vec{0}$ is to set all constants to zero (i.e., "use none of them"), then no vector cancels another. Every vector contributes a direction you cannot get from the rest. **Independent.**

The system of equations is just the mechanical tool to determine which case you are in.
